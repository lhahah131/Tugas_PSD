```python
import numpy as np


def konvolusi_diskrit(x, h):
    """Konvolusi 1D untuk dua sinyal diskrit berhingga."""
    return np.convolve(x, h)


def soal_1():
    # x[n] = 1 untuk 0<=n<3, x[3]=2, selain itu 0
    x = np.array([1, 1, 1, 2], dtype=float)
    h = np.array([1, 1, 2], dtype=float)
    y = konvolusi_diskrit(x, h)
    return x, h, y


def soal_2():
    x = np.array([1, 1, 1, 2], dtype=float)
    h = np.array([0, 0, 2, 1, 0], dtype=float)
    y = konvolusi_diskrit(x, h)
    return x, h, y


def y_teori_soal_3(n):
    # y[n] = (1/2)^n * (4^(n+1)-1)/3 untuk n>=0
    return (0.5 ** n) * ((4 ** (n + 1) - 1) / 3.0)


def y_numerik_soal_3(n_max=10):
    # Verifikasi numerik dari definisi konvolusi:
    # y[n] = sum_{k=0..n} 2^k * (1/2)^(n-k)
    y = []
    for n in range(n_max + 1):
        total = 0.0
        for k in range(n + 1):
            total += (2 ** k) * ((0.5) ** (n - k))
        y.append(total)
    return np.array(y, dtype=float)


def verifikasi_soal_3(n_max=10):
    n = np.arange(0, n_max + 1)
    y_num = y_numerik_soal_3(n_max)
    y_teo = y_teori_soal_3(n)
    cocok = np.allclose(y_num, y_teo, atol=1e-12)
    return n, y_num, y_teo, cocok


def main():
    np.set_printoptions(suppress=True)

    x1, h1, y1 = soal_1()
    print("Soal 1")
    print("x =", x1.astype(int))
    print("h =", h1.astype(int))
    print("y =", y1.astype(int))
    print()

    x2, h2, y2 = soal_2()
    print("Soal 2")
    print("x =", x2.astype(int))
    print("h =", h2.astype(int))
    print("y =", y2.astype(int))
    print()

    n, y_num, y_teo, cocok = verifikasi_soal_3(n_max=8)
    print("Soal 3")
    print("n      =", n)
    print("y_num  =", np.round(y_num, 6))
    print("y_teori=", np.round(y_teo, 6))
    print("cocok? =", cocok)


if __name__ == "__main__":
    main()
```

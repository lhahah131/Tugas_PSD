```python
import numpy as np 
import matplotlib.pyplot as plt 
 
#1.Generasi sinyal dengan fase acak
t = np.linspace(0, 1, 500)
A = 1 #amplitudo
f = 5 #Frekuensi

#menghasilkan sinyal sudut acak 
phi_acak = np.random.uniform(-1, 1)

y = A * np.sin(2 * np.pi * f * t + phi_acak)

#visualisasi
plt.plot(figsize=(10, 4))
plt.plot(t, y, color='blue')
plt.title(f'Sinyal dengan fase acak: {phi_acak:.4f}')
plt.xlabel('Waktu(t)')
plt.ylabel('Amplitudo (v)')
plt.grid(True)
plt.show()

print(f"Fase yang terpilihj secara acak adalah: {phi_acak} radian")
```

pada sebauh program yang diatas dan juga grafiknya itu untuk percobaan saya bahwa penggunaan nilai fase (phi)yang acak mengakibatkan sebauh pergeseran posisi horizontal pada sebauh globang amplitudo maupun frekuensinya.Sinyal akan tetap memiliki karekteristik gelombang yang sama,namun pada titip awalnya itu memulai tidak dapat di presdiksi setiap kali program dijalankan.

```python
import numpy as np              # Pastikan ada huruf 'p' di sini
import matplotlib.pyplot as plt

# 2. Generasi sinyal dengan frekuensi acak
t = np.linspace(0, 1, 500)      # Gunakan np.linspace, bukan n.linspace
A = 1                           # Amplitudo tetap
phi = 0                         # Fase tetap

# Menghasilkan frekuensi acak (distribusi uniform -1 hingga +1 Hz)
f_acak = np.random.uniform(-1, 1)

# Rumus sinyal
y = A * np.sin(2 * np.pi * f_acak * t + phi)

# Visualisasi
plt.figure(figsize=(10, 4))
plt.plot(t, y, color='red')
plt.title(f'Sinyal dengan Frekuensi Acak (f): {f_acak:.4f} Hz')
plt.xlabel('Waktu (t)')
plt.ylabel('Amplitudo (V)')
plt.grid(True)
plt.show()
```

Dalam sebauh kondisi yang diatas dan program saya distribusui uniform pada parameter frekuensi menujukan sebauh pengaruh terhadap sinyal kecepatan isolasi sinyal.Perubahan ini juga dapat membuktikan frekuensi secara acak antara -1 hingga +1 Hz dibuktikan dengan bahwa meskipun amplitudo akan tetap konstan,informasi atau kecepatan data dalam sinyal tersebut berubah secra drastis.Fenomena ini dapat di artikan menjadi sebauh frekuensi negatif yang akan muncul ketika mensimulasikan pergeseran 180 derjat secara mmatematik,yang sering ditemui dalam analisa sinyak sistem kendali.

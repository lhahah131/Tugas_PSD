```python
import numpy as np
import matplotlib.pyplot as plt
# 1. Persiapan Waktu
t = np.linspace(0, 1, 500)

# 2. Modifikasi Parameter Amplitudo
A1 = 2.0    # Amplitudo Fasa 1 
A2 = 4.0    # Amplitudo Fasa 2 
A3 = 6.0    # Amplitudo Fasa 3 
f = 5       # Frekuensi tetap 5 Hz
phi1, phi2, phi3 = 0, 120, 240 # Fase dalam derajat
```

Dalam program diatas adalah menunjukan sbeuh pase dalam sebauh sisten untuk melkaukan pengulan tiga sinyal sinuoidal yang memiliki frekuensi yang sama,namun masih-masing memiliki perbedaan sudah dalam fase sebesar 120 derajat.untuk saya pase kedua adalah melakukan sebauh pebuatan nilai yang berbeda yang bertujuan untuk pembeda nilai itu sendiri maka jika tinggi puncak dari ketiga gelombang tersebut tidak akan sama.

```python
# 3. Pembuatan Gelombang Konversi derajat ke radian 
x1 = A1 * np.sin(2 * np.pi * f * t + phi1 * np.pi / 180)
x2 = A2 * np.sin(2 * np.pi * f * t + phi2 * np.pi / 180)
x3 = A3 * np.sin(2 * np.pi * f * t + phi3 * np.pi / 180)
```

dDalam code diatas adalah saya melakukan sebauh perhitungan untuk mecari amplitudonya meskipun berbeda,titikpuncak biasnya setiap glombang akan tetap tidak muncul bersamaan.namun ada sebauh delay yang akan muncul diantara mereka.

```python
# 4. Visualisasi dalam Satu Plot
plt.figure(figsize=(12, 6))
plt.plot(t, x1, 'b', label=f'Fasa 1 (A={A1})')
plt.plot(t, x2, 'r', label=f'Fasa 2 (A={A2})')
plt.plot(t, x3, 'g', label=f'Fasa 3 (A={A3})')
# Pengaturan Grafik
plt.xlabel('Waktu (t)')
plt.ylabel('Amplitudo (V)')
plt.title('Sinyal Sinusoidal 3 Fasa dengan Amplitudo Berbeda')
plt.legend(loc='upper right')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(0, color='black', linewidth=1)
plt.show()
```

Dalam konteks ini adalah saya melakuan sebauh pengaturan visualisasi unntuk grafiknya dengan melihat sebauh sistem yang di sebauh sitem tidak seimbang dengan ini kondisi seperti ini sangan harus di hindari karena dapat menyebabkan sebauh getaran pada kabel dan menyebankan kerusan jadi ini membutikan bahwa saya dapat mengontrol tinggi redahnya daya tiap fase melalui parameter amlitudonya.

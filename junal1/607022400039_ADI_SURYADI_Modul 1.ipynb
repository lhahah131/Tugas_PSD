```python
import numpy as np
import matplotlib.pyplot as plt

# 1. MODIFIKASI PARAMETER
A = 10         # (Amplitudo)
f = 4          # (Frekuensi)
t = np.linspace(0, 1, 500)

y = A * np.sin(2 * np.pi * f * t)

plt.figure(figsize=(10, 4))
plt.plot(t, y, color='blue', linewidth=2)
plt.title(f'Grafik Nomor 1: Modifikasi (A={A}, f={f} Hz)')
plt.xlabel('Waktu (detik)')
plt.ylabel('Amplitudo')
plt.grid(True)
plt.axhline(0, color='black', linewidth=1)
plt.show()

print(f"Tips: Amplitudo {A} membuat tinggi gelombang mencapai {A}. Frekuensi {f} membuat ada {f} gelombang dalam 1 detik.")
```

1.pada sebuah nomor satu ini saya melaukan sebauh pengaturan untuk ini  dnegna mengatur amplitudonya sebasar nilai.Nilai ini adalah sebauh batas tertinggi udan terendah dari sebauh simpangan gelombang di mana sebuah gelombang akan dapat menyentuh angka batasnya 10 itu pada sumbu Y dan nilai lemahnya itu mendapatkan nilai -10.Dengan ini saya melakukan sebuah pengaturna untuk frekuensinya itu sendiri yaitu untuk saya adalah 4.frekuensi adalah sebauh seberapa cepat gelombang tersebur merambat dalam sebauh waktu tertu.Dari yang saya masukan adalah untuk sebauh frekuensinya adalah 4 dengan ini bisa di artikan juga untuk ini adalah dengan 4 kali gelombang itu begerak dari titik 0 samapai titik 0 juga. 

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter dasar
A, f = 5, 2
t = np.linspace(0, 1, 500)
y = A * np.sin(2 * np.pi * f * t)

# 2. ANALISIS SINYAL
peak_val = np.max(y)
peak_idx = np.where(y == peak_val)[0]
zero_crossings = np.where(np.diff(np.sign(y)))[0]

plt.figure(figsize=(10, 4))
plt.plot(t, y, color='purple', label='Sinyal Utama')

# Menandai hasil analisis pada grafik
plt.scatter(t[peak_idx], y[peak_idx], color='black', s=100, label='Titik Puncak', zorder=5)
plt.scatter(t[zero_crossings], y[zero_crossings], color='green', marker='x', s=150, label='Zero Crossing', zorder=5)

plt.title('Grafik Nomor 2: Deteksi Puncak & Zero Crossing')
plt.legend()
plt.grid(True)
plt.show()

print(f"Hasil Analisis: Nilai Puncak = {peak_val}, Jumlah Zero Crossing = {len(zero_crossings)} kali.")
```

2.Pada sebauh penggabaran diatas adalah menujukan sebuah kondisi yang melakuakn sebauh penghitungan untuk menganalisa sinyal gelombang dengan otomatis untuk menetukan titik-titik penting pada gelombang.Saya menggunakan sebauh parameter untuk malakukan perhitungan dasar sanya dengan amplitudo 5 dan frekuensi 2,dengan ini kita bisa melakukan pengamatan untuk dalam perogramnya itu akan mencari sebuah titik puncak dengan memindai seluruh data untuk menemukan nilai amplitudo tertiggi dengan begitu akan di tandai dengan sebauh titik puncaknya dengan warna hitam pada grafik.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameter sama agar adil dibandingkannya
A, f = 5, 2
t = np.linspace(0, 1, 500)

# 3. PERHITUNGAN SIN VS COS
y_sin = A * np.sin(2 * np.pi * f * t)
y_cos = A * np.cos(2 * np.pi * f * t)

plt.figure(figsize=(10, 4))
plt.plot(t, y_sin, label='Fungsi Sinus', color='blue')
plt.plot(t, y_cos, label='Fungsi Kosinus', color='red', linestyle='--')

plt.title('Grafik Nomor 3: Perbandingan Sinus vs Kosinus')
plt.xlabel('Waktu (detik)')
plt.ylabel('Amplitudo')
plt.legend()
plt.grid(True)
plt.axhline(0, color='black', linewidth=1)
plt.show()

print("Komentar: Terlihat Sinus mulai dari 0, sedangkan Kosinus mulai dari nilai puncak (90 derajat lebih awal).")
```

Dalan nomor ini adalah saya melakukan sebuah di dalam program saya untuk melakukan pekerjaan membandingkan karakteristik dari bentuk sinus dan gelombang Kosinus secara langsung dalam satu grafik.Dalam ini saya menjelaskan untuk grafiknya untuk warnaya sinus di tandai dengan warna biru tebal,untuk ini selalu mulai penghitunganya itu dari titik 0 pada sumbu Y,lalu akan bergerak menuju ke puncak sebaliknya untuk sinus adalah warna merah putus-putus,yang tidak bergerak dari nol melainkan sudah berada pada titik puncak tertinggi di anggka 5 sejak detik ke-0.Dengan ini kita bisa memastikan seabuh siklus gelombang konsinus pada dasarnya adalaj gelombang sinus yang bergerak mendahuluinya dengan lebih awal dan bisa juga itu disebaut dia bergeser sekitar 90 derajat lebih awal melintasi waktunya.

```python
# --- 4. VISUALISASI ---
plt.figure(figsize=(12, 6))

# Plot Utama (Sinus)
plt.plot(t, y_sin, label=f'Sinus (A={A}, f={f})', color='blue', linewidth=2)

# Plot Pembanding (Kosinus)
plt.plot(t, y_cos, label='Kosinus', color='red', linestyle='--', alpha=0.5)

# Tandai Titik Puncak (Titik Hitam)
plt.scatter(t[peak_idx], y_sin[peak_idx], color='black', zorder=5, label='Titik Puncak')

# Tandai Zero Crossing (Tanda X Hijau)
plt.scatter(t[zero_crossings], y_sin[zero_crossings], color='green', marker='x', s=100, zorder=5, label='Zero Crossing')

# Pengaturan Tampilan
plt.title('Analisis Gelombang Sinus vs Kosinus', fontsize=14)
plt.xlabel('Waktu (detik)')
plt.ylabel('Amplitudo')
plt.axhline(0, color='black', linewidth=1) # Garis nol
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.show()
```

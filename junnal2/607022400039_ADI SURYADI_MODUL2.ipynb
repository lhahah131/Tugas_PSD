## 1. Jurnal I: Analisis Sinyal Eksponensial Real


```python
t = np.linspace(0, 2, 400)
a_vals = [1, 2, 3]   # Growing factors
b_vals = [-1, -2, -3] # Decaying factors

plt.figure(figsize=(12, 5))

# Plot Growing Exponential
plt.subplot(1, 2, 1)
for a in a_vals:
    plt.plot(t, np.exp(a * t), label=rf'$\alpha = {a}$')
plt.title('Pertumbuhan Eksponensial (Growing)')
plt.xlabel('Waktu (t)')
plt.ylabel('Amplitudo')
plt.legend()

# Plot Decaying Exponential
plt.subplot(1, 2, 2)
for b in b_vals:
    plt.plot(t, np.exp(b * t), label=rf'$\alpha = {b}$')
plt.title('Peluruhan Eksponensial (Decaying)')
plt.xlabel('Waktu (t)')
plt.ylabel('Amplitudo')
plt.legend()

plt.tight_layout()
plt.show()
```

pada saat α positif maka nilai yang naik sangat cepat dan seiring waktu semakin besar α maka akan semakin curan untuk kenaikanya sedangkan ketika negatif sinyalnya turun mendekati nol lalu ketikan kecil nilai α maka semakin cepat sinyal habis atau meluruh.

## 2. Jurnal II: Eksponensial Kompleks - Respon Magnitude dan Phase
Sinyal kompleks $x(t) = e^{j\\Omega t}$ memiliki komponen:
- **Magnitude**: $|x(t)| = 1$ (selalu bernilai 1)
- **Phase**: $\\phi(t) = \\Omega t$ (berubah linier terhadap waktu)

```python
t = np.linspace(0, 1, 500)
f = 10
Omega = 2 * np.pi * f
x = np.exp(1j * Omega * t)

mag_x = np.abs(x)
phase_x = np.angle(x)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(t, mag_x, color='blue', lw=2)
plt.ylim(0, 1.5)
plt.title(r'Respon Magnitude $|e^{j\Omega t}|$')
plt.xlabel('Waktu (s)')
plt.ylabel('Magnitude')

plt.subplot(1, 2, 2)
plt.plot(t, phase_x, color='green')
plt.title('Respon Phase (Wrapped)')
plt.xlabel('Waktu (s)')
plt.ylabel('Phase (radians)')

plt.tight_layout()
plt.show()
```

Untuk yang nomor 2 dalah memabahas sebauh magnitude selalu 1 aratinya sinyal tidak berubahh terhadap waktu,jadi grafiknya magnitude berupa garis datar di angak 1 sedangkan phase seiring waktu sudutnya bertambah linier saat waktu bertambah dalam kasus ini kenapa sebauh grafiknya naik turun karena tang dipakai adalah wrapped phase jadi saat melewati batas nilainya lompat kembali.

## 3. Jurnal III: Sinyal Satuan Diskrit (Unit Step & Unit Impulse)


```python
n = np.arange(-10, 11)

unit_step = (n >= 0).astype(float)
unit_impulse = (n == 0).astype(float)
unit_ramp = n * unit_step

plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.stem(n, unit_impulse)
plt.title('Unit Impulse $\delta[n]$')

plt.subplot(1, 3, 2)
plt.stem(n, unit_step)
plt.title('Unit Step $u[n]$')

plt.subplot(1, 3, 3)
plt.stem(n, unit_ramp)
plt.title('Unit Ramp $r[n]$')

plt.tight_layout()
plt.show()
```

Unit impulse δ[n]: hanya bernilai 1 di n=0, selain itu 0.
Unit step u[n]: bernilai 0 untuk n<0, lalu 1 untuk n≥0.
Unit ramp r[n]: bernilai 0 untuk n<0, lalu naik linier untuk n≥0.
Intinya, tiga sinyal ini sering jadi “bahan dasar” untuk membangun sinyal lain.

## 4. Jurnal IV: Pembangkitan Waveform Dasar (Square & Sawtooth Wave)


```python
t_wf = np.linspace(0, 1, 1000)
f_wf = 5

plt.figure(figsize=(12, 8))

# Square Wave
plt.subplot(2, 2, 1)
plt.plot(t_wf, signal.square(2 * np.pi * f_wf * t_wf))
plt.title('Square Wave (5 Hz)')

# Sawtooth Wave
plt.subplot(2, 2, 2)
plt.plot(t_wf, signal.sawtooth(2 * np.pi * f_wf * t_wf))
plt.title('Sawtooth Wave (5 Hz)')

# Sinc Function
t_sinc = np.linspace(-10, 10, 1000)
plt.subplot(2, 2, 3)
plt.plot(t_sinc, np.sinc(t_sinc))
plt.title('Sinc Function')

# Gaussian Pulse
plt.subplot(2, 2, 4)
plt.plot(t_sinc, np.exp(-t_sinc**2 / 2))
plt.title('Gaussian Pulse')

plt.tight_layout()
plt.show()
```

Square wave: bentuk kotak, naik turun tajam.
Sawtooth wave: bentuk gigi gergaji, naik/turun linear lalu reset.
Sinc function: puncak di tengah dengan osilasi mengecil ke samping.
Gaussian pulse: kurva lonceng halus.

## 5. Jurnal V: Modulasi Amplitudo pada Sinyal Suara


```python
# Referensi Frekuensi Nada Dasar
notes = {
    'C': 261.63, 'D': 293.66, 'E': 329.63, 'F': 349.23, 
    'G': 392.00, 'A': 440.00, 'B': 493.88, 'C5': 523.25
}

fs = 8000
duration = 0.4
t_note = np.linspace(0, duration, int(fs * duration), endpoint=False)

def generate_tone(freq):
    return 0.5 * np.sin(2 * np.pi * freq * t_note)

# Melodi: C - E - G - C5
melody_freqs = [notes['C'], notes['E'], notes['G'], notes['C5']]
song = np.concatenate([generate_tone(f) for f in melody_freqs])

# Menampilkan Spektrogram nada terakhir sebagai visualisasi
plt.figure(figsize=(10, 3))
plt.specgram(song, Fs=fs, cmap='viridis')
plt.title('Spektrogram Melodi C-E-G-C5')
plt.ylabel('Frekuensi (Hz)')
plt.xlabel('Waktu (s)')
plt.colorbar(label='Intensity (dB)')
plt.show()

print("Eksperimen selesai. Gunakan sd.play(song, fs) untuk mendengarkan hasil.")
```

Module amplitude pada sebauh sinyal suara bagian ini membeut nada naik sederhana dari prekuensi  tertrntu (C, E, G, C5),lalu jadi digabungkan mejadi suara melodi setelah itu ditampilkan spektrogram untuk melihat sebauh prekuensinya.

## 6. Jurnal VI: Aplikasi - Visualisasi Sinyal Audio (.wav)


```python
def plot_wav(filename, title, color):
    if os.path.exists(filename):
        fs, data = wavfile.read(filename)
        # Jika stereo, ambil satu channel
        if len(data.shape) > 1:
            data = data[:, 0]
        t_audio = np.linspace(0, len(data)/fs, len(data))
        plt.plot(t_audio, data, color=color, alpha=0.7)
        plt.title(title)
        plt.xlabel('Waktu (s)')
        plt.ylabel('Amplitudo')
    else:
        plt.text(0.5, 0.5, f'File {filename}\ntidak ditemukan', ha='center', va='center')
        plt.title(title)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plot_wav('audio_naik.wav', 'Visualisasi audio_naik.wav', 'blue')

plt.subplot(1, 2, 2)
plot_wav('audio_turun.wav', 'Visualisasi audio_turun.wav', 'red')

plt.tight_layout()
plt.show()
```

Jika audionya stereo, diambil 1 channel saja.
Jika file tidak ada, muncul pesan file tidak ditemukan.
Intinya, ini latihan membaca data audio nyata dan melihat pola amplitudonya.

## 7. Jurnal VII: Make a Music - Sintesis Melodi Sederhana


```python
import numpy as np
import sounddevice as sd
import matplotlib.pyplot as plt

# Parameter audio
fs = 8000  # Sampling rate
duration = 0.5  # Durasi per nada (detik)
t = np.linspace(0, duration, int(fs * duration), endpoint=False)

# Definisi frekuensi nada (Hz)
notes = {
    'C4': 261.63, 'D4': 293.66, 'E4': 329.63, 'F4': 349.23,
    'G4': 392.00, 'A4': 440.00, 'B4': 493.88, 'C5': 523.25
}

def generate_note(freq, t):
    # Menghasilkan sinyal sinus untuk frekuensi tertentu
    return np.sin(2 * np.pi * freq * t)

# Membuat melodi: C4 - D4 - E4 - F4 - G4 - A4 - B4 - C5
melody_keys = ['C4', 'D4', 'E4', 'F4', 'G4', 'A4', 'B4', 'C5']
audio_sequence = []

for key in melody_keys:
    tone = generate_note(notes[key], t)
    audio_sequence.append(tone)

# Menggabungkan semua nada menjadi satu array
full_melody = np.concatenate(audio_sequence)

print("Memainkan Melodi...")
sd.play(full_melody, fs)
sd.wait()  # Tunggu sampai selesai
print("Selesai.")

# Visualisasi sebagian kecil dari melodi
plt.figure(figsize=(12, 4))
plt.plot(full_melody[:2000])
plt.title('Visualisasi Sinyal Melodi (Awal)')
plt.xlabel('Sample Index')
plt.ylabel('Amplitudo')
plt.show()
```

Membuat sinyal sinus per nada.
Menggabungkan semua nada jadi satu sinyal panjang.
Memutar hasil audio dan menampilkan sebagian bentuk sinyalnya.
Intinya, ini praktik sintesis suara sederhana: dari frekuensi jadi musik.

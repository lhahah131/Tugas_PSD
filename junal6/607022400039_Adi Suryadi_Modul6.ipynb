```python
# Nomor 1 - Amplitude scaling (impulse)
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-10, 11, 1)
x = (n == 0).astype(int)

A = [2, 0.5, -1]
y1 = A[0] * x
y2 = A[1] * x
y3 = A[2] * x

plt.figure(figsize=(10, 7))

plt.subplot(2, 2, 1)
plt.stem(n, x, linefmt='C0-', markerfmt='C0o', basefmt='k-')
plt.title('x[n]')
plt.grid(alpha=0.25)

plt.subplot(2, 2, 2)
plt.stem(n, y1, linefmt='C2-', markerfmt='C2o', basefmt='k-')
plt.title('y1[n] = 2x[n]')
plt.grid(alpha=0.25)

plt.subplot(2, 2, 3)
plt.stem(n, y2, linefmt='C1-', markerfmt='C1o', basefmt='k-')
plt.title('y2[n] = 0.5x[n]')
plt.grid(alpha=0.25)

plt.subplot(2, 2, 4)
plt.stem(n, y3, linefmt='C3-', markerfmt='C3o', basefmt='k-')
plt.title('y3[n] = -x[n]')
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

**Keterangan Grafik (Nomor 1 - Impuls)**
- Impuls hanya bernilai tidak nol di n = 0.
- Pengali amplitudo hanya mengubah tinggi batang impuls, tidak mengubah posisi waktunya.
- Kasus A negatif menghasilkan impuls terbalik terhadap sumbu horizontal.

```python
# Nomor 2 - Amplitudo shifting (sinus)
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1, 100)
x = np.sin(2 * np.pi * 5 * t)
offset = [0.5, -0.5]

y1 = x + offset[0]
y2 = x + offset[1]

plt.figure(figsize=(9, 8))
plt.subplot(3, 1, 1)
plt.plot(t, x, color='teal')
plt.title('x[n]')
plt.grid(alpha=0.25)

plt.subplot(3, 1, 2)
plt.plot(t, y1, color='darkorange')
plt.title('y1[n] = x[n] + 0.5')
plt.grid(alpha=0.25)

plt.subplot(3, 1, 3)
plt.plot(t, y2, color='crimson')
plt.title('y2[n] = x[n] - 0.5')
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

**Keterangan Grafik (Nomor 2 - Offset Sinus)**
- Grafik tengah menunjukkan efek offset positif (+0.5), grafik bawah offset negatif (-0.5).
- Seluruh kurva berpindah vertikal tanpa perubahan periode.
- Operasi ini merepresentasikan penambahan level DC pada sinyal.

```python
# Nomor 4 - Perkalian dua sinyal
import numpy as np
import matplotlib.pyplot as plt

t = np.arange(0, 100, 1)
Fs = 100

x1 = np.sin(2 * np.pi * (5 / Fs) * t)
x2 = np.cos(2 * np.pi * (5 / Fs) * t)
y = x1 * x2

plt.figure(figsize=(10, 8))
plt.subplot(3, 1, 1)
plt.stem(t, x1, linefmt='C0-', markerfmt='C0o', basefmt='k-')
plt.title('x1[n]')

plt.subplot(3, 1, 2)
plt.stem(t, x2, linefmt='C4-', markerfmt='C4o', basefmt='k-')
plt.title('x2[n]')

plt.subplot(3, 1, 3)
plt.stem(t, y, linefmt='C2-', markerfmt='C2o', basefmt='k-')
plt.title('y[n] = x1[n] * x2[n]')

plt.tight_layout()
plt.show()
```

**Keterangan Grafik (Perkalian x1 dan x2)**
- Dua sinyal masukan (sinus dan cosinus) dikalikan titik per titik.
- Hasil y[n] memperlihatkan perubahan bentuk amplitudo akibat interaksi kedua sinyal.
- Ini adalah contoh dasar operasi modulasi pada domain waktu diskrit.

```python
# Nomor 4 - Penjumlahan sinyal dan sifat komutatif
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-10, 11)
x = (n >= 0).astype(int)
y = (n >= 2).astype(int)

z1 = x + y
z2 = y + x

plt.figure(figsize=(9, 7))
plt.subplot(3, 1, 1)
plt.stem(n, x, linefmt='C0-', markerfmt='C0o', basefmt='k-')
plt.title('x[n]')

plt.subplot(3, 1, 2)
plt.stem(n, y, linefmt='C1-', markerfmt='C1o', basefmt='k-')
plt.title('y[n]')

plt.subplot(3, 1, 3)
plt.stem(n, z1, linefmt='C3-', markerfmt='C3o', basefmt='k-')
plt.title('x[n] + y[n]')

plt.tight_layout()
plt.show()

print('Komutatif:', np.array_equal(z1, z2))
```

**Keterangan Grafik (Penjumlahan dan Komutatif)**
- Grafik ketiga adalah hasil x[n] + y[n].
- Dari perhitungan, x[n] + y[n] sama dengan y[n] + x[n].
- Hasil print Komutatif: True memverifikasi sifat komutatif.

```python
# Nomor 5 - Upsampling dan downsampling
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 10)
x = np.array([1, 2, 3, 4, 5, 4, 3, 2, 1, 0])

L = 2
y = np.zeros(len(x) * L)
y[::L] = x
z = y[::L]

plt.figure(figsize=(9, 7))
plt.subplot(3, 1, 1)
plt.stem(n, x, linefmt='C0-', markerfmt='C0o', basefmt='k-')
plt.title('x[n] (original)')

plt.subplot(3, 1, 2)
plt.stem(np.arange(len(y)), y, linefmt='C2-', markerfmt='C2o', basefmt='k-')
plt.title('Upsampling')

plt.subplot(3, 1, 3)
plt.stem(np.arange(len(z)), z, linefmt='C4-', markerfmt='C4o', basefmt='k-')
plt.title('Downsampling kembali')

plt.tight_layout()
plt.show()
```

**Keterangan Grafik (Upsampling dan Downsampling)**
- Upsampling menyisipkan nol di antara sampel asli sehingga laju sampel naik.
- Downsampling kembali mengambil sampel tiap L sehingga bentuk utama sinyal kembali.
- Pada contoh ini, pola sinyal asli tetap dapat dikenali setelah proses naik-turun sampel.

```python
# Nomor 7 - Delay dan advance
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-10, 11)
x = (n >= 0).astype(int)
k = 3

delay = (n >= k).astype(int)
advance = (n >= -k).astype(int)

plt.figure(figsize=(9, 7))
plt.subplot(3, 1, 1)
plt.stem(n, x, linefmt='C0-', markerfmt='C0o', basefmt='k-')
plt.title('x[n]')

plt.subplot(3, 1, 2)
plt.stem(n, delay, linefmt='C3-', markerfmt='C3o', basefmt='k-')
plt.title('Delay x[n-3]')

plt.subplot(3, 1, 3)
plt.stem(n, advance, linefmt='C2-', markerfmt='C2o', basefmt='k-')
plt.title('Advance x[n+3]')

plt.tight_layout()
plt.show()
```

**Keterangan Grafik (Delay dan Advance)**
- Delay x[n-k] menggeser sinyal ke kanan (terlambat).
- Advance x[n+k] menggeser sinyal ke kiri (lebih cepat).
- Bentuk step tetap sama, hanya posisi terhadap indeks n yang berubah.

```python
# Bagian konvolusi: rectangular, shifted impulse, dan sifat konvolusi
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-5, 6)
x = ((n >= -2) & (n <= 2)).astype(int)
h = (n == 0).astype(int)
y = np.convolve(x, h)
n_conv = np.arange(n[0] + n[0], n[-1] + n[-1] + 1)

plt.figure(figsize=(6, 8))
plt.subplot(3, 1, 1)
plt.stem(n, x, linefmt='C0-', markerfmt='C0o', basefmt='k-')
plt.title('Rectangular x[n]')
plt.xlabel('n')

plt.subplot(3, 1, 2)
plt.stem(n, h, linefmt='C2-', markerfmt='C2o', basefmt='k-')
plt.title('Impulse h[n]')
plt.xlabel('n')

plt.subplot(3, 1, 3)
plt.stem(n_conv, y, linefmt='C3-', markerfmt='C3o', basefmt='k-')
plt.title('y[n] = x[n] * h[n]')
plt.xlabel('n')
plt.tight_layout()
plt.show()

k = 3
h1 = (n == k).astype(int)
h2 = (n == -k).astype(int)
y1 = np.convolve(x, h1)
y2 = np.convolve(x, h2)

plt.figure(figsize=(8, 8))
plt.subplot(3, 2, 1)
plt.stem(n, x, linefmt='C0-', markerfmt='C0o', basefmt='k-'); plt.title('x[n]')
plt.subplot(3, 2, 2)
plt.stem(n, x, linefmt='C0-', markerfmt='C0o', basefmt='k-'); plt.title('x[n]')
plt.subplot(3, 2, 3)
plt.stem(n, h1, linefmt='C4-', markerfmt='C4o', basefmt='k-'); plt.title('δ[n-3]')
plt.subplot(3, 2, 4)
plt.stem(n, h2, linefmt='C1-', markerfmt='C1o', basefmt='k-'); plt.title('δ[n+3]')
plt.subplot(3, 2, 5)
plt.stem(n_conv, y1, linefmt='C3-', markerfmt='C3o', basefmt='k-'); plt.title('y1[n] (delay)')
plt.subplot(3, 2, 6)
plt.stem(n_conv, y2, linefmt='C2-', markerfmt='C2o', basefmt='k-'); plt.title('y2[n] (advance)')
plt.tight_layout()
plt.show()

x_prop = np.array([1, 2, 3])
h_prop = np.array([1, 1, 1])
g_prop = np.array([2, 1, 0])

print('Komutatif:', np.array_equal(np.convolve(x_prop, h_prop), np.convolve(h_prop, x_prop)))
print('Asosiatif:', np.allclose(np.convolve(np.convolve(x_prop, h_prop), g_prop), np.convolve(x_prop, np.convolve(h_prop, g_prop))))
left = np.convolve(x_prop, h_prop + g_prop)
right = np.convolve(x_prop, h_prop) + np.convolve(x_prop, g_prop)
print('Distributif:', np.allclose(left, right))
```

**Keterangan Grafik (Konvolusi dan Sifat Konvolusi)**
- Konvolusi x[n] dengan delta[n] menghasilkan sinyal yang sama dengan x[n].
- Jika impuls digeser, hasil konvolusi ikut bergeser (delay atau advance).
- Nilai True pada output membuktikan sifat komutatif, asosiatif, dan distributif konvolusi pada contoh ini.

```python
# Nomor 8 - Kombinasi beberapa sinyal impuls
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-5, 6)
x1 = (n == -1).astype(int) + (n == 1).astype(int)
x2 = (n == -1).astype(int) - (n == 1).astype(int)
x3 = (n == 0).astype(int) + 2 * (n == 1).astype(int) + (n == 2).astype(int)
x4 = (n == 0).astype(int) - (n == 1).astype(int) + (n == 2).astype(int)

plt.figure(figsize=(8, 6))
plt.subplot(4, 1, 1); plt.stem(n, x1, linefmt='C0-', markerfmt='C0o', basefmt='k-'); plt.title('x1')
plt.subplot(4, 1, 2); plt.stem(n, x2, linefmt='C1-', markerfmt='C1o', basefmt='k-'); plt.title('x2')
plt.subplot(4, 1, 3); plt.stem(n, x3, linefmt='C2-', markerfmt='C2o', basefmt='k-'); plt.title('x3')
plt.subplot(4, 1, 4); plt.stem(n, x4, linefmt='C3-', markerfmt='C3o', basefmt='k-'); plt.title('x4')
plt.tight_layout()
plt.show()
```

**Keterangan Grafik (Kombinasi Sinyal Impuls)**
- Setiap subplot menunjukkan kombinasi letak dan koefisien impuls yang berbeda.
- Nilai koefisien positif/negatif mempengaruhi arah amplitudo batang impuls.
- Visualisasi ini membantu memahami representasi sinyal diskrit berbasis impuls.

```python
# Autocorrelation dan cross-correlation
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 2, 1])
y = np.array([0, 1, 0, 1, 0])

rxx = np.correlate(x, x, mode='full')
rxy = np.correlate(x, y, mode='full')
lag = np.arange(-len(x) + 1, len(x))

plt.figure(figsize=(8, 6))
plt.subplot(2, 1, 1)
plt.stem(lag, rxx, linefmt='C0-', markerfmt='C0o', basefmt='k-')
plt.title('Autocorrelation')

plt.subplot(2, 1, 2)
plt.stem(lag, rxy, linefmt='C5-', markerfmt='C5o', basefmt='k-')
plt.title('Cross-correlation')

plt.tight_layout()
plt.show()
```

**Keterangan Grafik (Autocorrelation dan Cross-correlation)**
- Autocorrelation bernilai maksimum saat lag = 0 karena sinyal dibandingkan dengan dirinya sendiri.
- Cross-correlation menunjukkan kemiripan dua sinyal terhadap pergeseran waktu.
- Puncak korelasi yang lebih besar menandakan kemiripan yang lebih kuat pada lag tertentu.

## Keterangan Hasil dan Pembahasan
### Keterangan per Percobaan
1. Amplitude Scaling (impulse)
   - Saat pengali A = 2, amplitudo impuls membesar dua kali.
   - Saat A = 0.5, amplitudo mengecil setengahnya.
   - Saat A = -1, amplitudo dibalik terhadap sumbu horizontal.
2. Amplitude Shifting (sinus)
   - Penambahan offset +0.5 menggeser sinyal ke atas.
   - Penambahan offset -0.5 menggeser sinyal ke bawah.
   - Bentuk sinyal tetap, hanya level DC yang berubah.
3. Perkalian dua sinyal sinus-cosinus
   - Hasil perkalian membentuk modulasi amplitudo sesuai identitas trigonometri.
   - Nilai keluaran dipengaruhi oleh fase dan frekuensi kedua sinyal.
4. Penjumlahan sinyal step (komutatif)
   - Grafik x[n] + y[n] sama dengan y[n] + x[n].
   - Verifikasi numerik menunjukkan sifat komutatif bernilai benar.
5. Upsampling dan downsampling
   - Upsampling faktor 2 menambah sampel nol di antara data asli.
   - Downsampling kembali dengan faktor sama mengembalikan bentuk data utama.
6. Delay dan advance
   - Delay x[n-k] menggeser sinyal ke kanan.
   - Advance x[n+k] menggeser sinyal ke kiri.
7. Konvolusi dengan impuls dan impuls geser
   - Konvolusi dengan delta[n] menghasilkan sinyal asal.
   - Konvolusi dengan delta[n-k] memberi efek delay, sedangkan delta[n+k] memberi efek advance.
   - Uji komutatif, asosiatif, dan distributif pada konvolusi menunjukkan hasil benar.
8. Kombinasi sinyal impuls
   - Tiap kombinasi menunjukkan posisi dan bobot impuls yang berbeda.
   - Perubahan koefisien langsung mengubah bentuk sinyal diskrit.
9. Autocorrelation dan cross-correlation
   - Autocorrelation maksimum saat lag = 0 karena sinyal dibandingkan dengan dirinya sendiri.
   - Cross-correlation menunjukkan tingkat kemiripan dua sinyal pada pergeseran tertentu.
